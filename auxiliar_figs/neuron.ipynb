```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle, FancyArrowPatch, PathPatch
from matplotlib.path import Path

plt.rcParams.update({
    "font.size": 18,
    "axes.titlesize": 18,
    "axes.labelsize": 18,
    "mathtext.fontset": "cm",
    "font.family": "serif"
})

fig, ax = plt.subplots(figsize=(14, 7))

ax.set_xlim(0, 14)
ax.set_ylim(0, 8)
ax.axis("off")

input_x = 2.7
input_y = [6.2, 4.2, 2.0]
input_labels = [r"$x_1$", r"$x_2$", r"$x_n$"]

sum_x, sum_y = 7.0, 4.2
sum_r = 0.78

act_x, act_y = 9.45, 4.2
act_w, act_h = 1.15, 0.90

out_x, out_y = 12.55, 4.2
bias_x, bias_y = 7.0, 7.05

def add_arrow(start, end, lw=1.7, ms=16):
    ax.add_patch(
        FancyArrowPatch(
            start,
            end,
            arrowstyle="->",
            mutation_scale=ms,
            linewidth=lw,
            color="black",
            shrinkA=0,
            shrinkB=0
        )
    )

def add_line(start, end, lw=1.7):
    ax.add_patch(
        FancyArrowPatch(
            start,
            end,
            arrowstyle="-",
            linewidth=lw,
            color="black",
            shrinkA=0,
            shrinkB=0
        )
    )

brace_x = 1.05
brace_top = 6.55
brace_bottom = 1.65
brace_mid = input_y[1]

brace_path = Path(
    [
        (brace_x, brace_top),
        (0.82, brace_top), (0.76, 6.30), (0.76, 5.95),
        (0.76, 5.20), (0.62, 4.75), (0.45, 4.48),
        (0.62, 4.32), (0.76, 3.95), (0.76, 3.25),
        (0.76, 2.50), (0.82, brace_bottom), (brace_x, brace_bottom),
    ],
    [
        Path.MOVETO,
        Path.CURVE4, Path.CURVE4, Path.CURVE4,
        Path.CURVE4, Path.CURVE4, Path.CURVE4,
        Path.CURVE4, Path.CURVE4, Path.CURVE4,
        Path.CURVE4, Path.CURVE4, Path.CURVE4,
    ]
)

ax.add_patch(
    PathPatch(
        brace_path,
        facecolor="none",
        edgecolor="black",
        linewidth=1.8
    )
)

#AQUI INPUTS
ax.text(-1.25, brace_mid, "Inputs", ha="left", va="center", fontsize=28)

for xlab, y in zip(input_labels, input_y):
    ax.text(input_x - 1.05, y, xlab, ha="right", va="center", fontsize=28)

    ax.add_patch(
        Circle(
            (input_x, y),
            radius=0.13,
            edgecolor="black",
            facecolor="white",
            linewidth=1.8
        )
    )

add_line((input_x + 0.18, input_y[0]), (sum_x - sum_r, sum_y), lw=1.6)
add_arrow((input_x + 0.18, input_y[1]), (sum_x - sum_r, sum_y), lw=1.6, ms=16)
add_line((input_x + 0.18, input_y[2]), (sum_x - sum_r, sum_y), lw=1.6)

ax.text(input_x - 0.70, 3.05, r"$\vdots$", ha="center", va="center", fontsize=34)

weights_x = 4.50

ax.text(weights_x, 5.55, r"$w_1$", ha="center", va="center", fontsize=26)
ax.text(weights_x, 4.45, r"$w_2$", ha="center", va="center", fontsize=26)
ax.text(weights_x, 3.5, r"$w_n$", ha="center", va="center", fontsize=26)

ax.text(weights_x, 1.05, "Weights", ha="center", va="center", fontsize=30)

ax.add_patch(
    Circle(
        (bias_x, bias_y),
        radius=0.13,
        edgecolor="black",
        facecolor="white",
        linewidth=1.8
    )
)

ax.text(bias_x, bias_y + 0.98, "Bias", ha="center", va="center", fontsize=26)
ax.text(bias_x, bias_y + 0.52, r"$b$", ha="center", va="center", fontsize=28)

add_arrow((bias_x, bias_y - 0.18), (sum_x, sum_y + sum_r), lw=1.6)

ax.add_patch(
    Circle(
        (sum_x, sum_y),
        radius=sum_r,
        edgecolor="black",
        facecolor="white",
        linewidth=1.9
    )
)

ax.text(sum_x, sum_y - 0.01, r"$\sum$", ha="center", va="center", fontsize=36)

add_arrow((sum_x + sum_r, sum_y), (act_x - act_w / 2, act_y), lw=1.7)

ax.add_patch(
    Rectangle(
        (act_x - act_w / 2, act_y - act_h / 2),
        width=act_w,
        height=act_h,
        edgecolor="black",
        facecolor="white",
        linewidth=1.9
    )
)

ax.text(act_x, act_y, r"$f$", ha="center", va="center", fontsize=34)
ax.text(act_x, act_y + 1.35, "Activation\nFunction", ha="center", va="center", fontsize=24)

add_arrow((act_x + act_w / 2, act_y), (out_x - 0.70, out_y), lw=1.7)

ax.text(out_x, out_y + 0.42, "Output", ha="center", va="center", fontsize=28)
ax.text(out_x, out_y - 0.20, r"$\hat{x}_{t+1}$", ha="center", va="center", fontsize=30)

fig.savefig(
    "single_neuron_scheme.pdf",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

fig.savefig(
    "single_neuron_scheme.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

print("Figura salva como single_neuron_scheme.pdf e single_neuron_scheme.png")
```
